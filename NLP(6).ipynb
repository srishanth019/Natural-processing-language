```python
import gensim.downloader as api


glove = api.load("glove-wiki-gigaword-50")


word = "computer"
vector = glove[word]

print("Word:", word)
print("Vector:", vector)


similar = glove.most_similar(word, topn=5)

print("\nSimilar words:")
for w, score in similar:
    print(w, ":", round(score, 3))
```
